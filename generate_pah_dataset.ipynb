# Benzene

```python
# |  F   |  Cl   |   O   |   S   |  C=O  |   OC   | C(O)=O |   N   |  C#N  |  C#C  |  C=C   |   C   |  CC   |  CCC  | N(=O)O | S(=O)(=O)O | C(N)=O | C(=O)C |
# | C(Cl)(Cl)Cl | C(Cl)Cl |  CCl   | C(F)(F)F | C(F)F |  CF   | C1=CC=CC=C1 |
from rdkit import Chem
from rdkit.Chem import Draw

substructures = ['F', 'Cl', 'O', 'S', 'C=O', 'OC', 'C(O)=O', 'N', 'C#N', 'C#C', 'C=C', 'C', 'CC', 'CCC', '[N+](=O)[O-]', 'S(=O)(=O)O', 'P(O)(O)=O', 'C(N)=O', 'C(=O)C', 'C(Cl)(Cl)Cl', 'C(Cl)Cl', 'CCl', 'C(F)(F)F', 'C(F)F', 'CF']
len(substructures)
```

```python
import matplotlib.pyplot as plt
# single substitution
single_substitution = []
substitute_type = []
substitute_substructure = []

for i, substructure in enumerate(substructures):
    tmp_smiles = 'C1=CC=CC=C1' + substructure
    mol = Chem.MolFromSmiles(tmp_smiles)
    single_substitution.append(Chem.MolToSmiles(mol, canonical=True, isomericSmiles=True, kekuleSmiles=True))
    substitute_type.append('1')
    substitute_substructure.append(substructure)
```

```python
# neighbor substitution
ortho_substitution = []
ortho_substitution_type = []
ortho_substitution_substructure = []

for substructure_left in substructures:
    for substructure_right in substructures:
        tmp_smiles = f'C1=CC=CC({substructure_left})=C1{substructure_right}'
        mol = Chem.MolFromSmiles(tmp_smiles)
        ortho_substitution.append(Chem.MolToSmiles(mol, canonical=True, isomericSmiles=False, kekuleSmiles=True))
        ortho_substitution_type.append('1,2')
        ortho_substitution_substructure.append([substructure_left, substructure_right])

        
```

```python
# meta substitution
meta_substitution = []
meta_substitution_type = []
meta_substitution_substructure = []

for substructure_left in substructures:
    for substructure_right in substructures:
        tmp_smiles = f'C1=CC=C({substructure_left})C=C1{substructure_right}'
        mol = Chem.MolFromSmiles(tmp_smiles)
        meta_substitution.append(Chem.MolToSmiles(mol, canonical=True, isomericSmiles=False, kekuleSmiles=True))
        meta_substitution_type.append('1,3')
        meta_substitution_substructure.append([substructure_left, substructure_right])


```

```python
# para substitution
para_substitution = []
para_substitution_type = []
para_substitution_substructure = []

for substructure_left in substructures:
    for substructure_right in substructures:
        tmp_smiles = f'C1=CC({substructure_left})=CC=C1{substructure_right}'
        mol = Chem.MolFromSmiles(tmp_smiles)
        para_substitution.append(Chem.MolToSmiles(mol, canonical=True, isomericSmiles=False, kekuleSmiles=True))
        para_substitution_type.append('1,4')
        para_substitution_substructure.append([substructure_left, substructure_right])

```

```python
benzene_substitution = single_substitution + ortho_substitution + meta_substitution + para_substitution
benzene_substitution_type = substitute_type + ortho_substitution_type + meta_substitution_type + para_substitution_type
benzene_substitution_substructure = substitute_substructure + ortho_substitution_substructure + meta_substitution_substructure + para_substitution_substructure
(len(benzene_substitution), len(benzene_substitution_type), len(benzene_substitution_substructure))
```

```python
# get optimal 3d conformations and output xyz files

from rdkit.Chem import AllChem

for i, smiles in enumerate(benzene_substitution):
    mol = Chem.MolFromSmiles(smiles)
    mol = Chem.AddHs(mol)
    AllChem.EmbedMolecule(mol)
    AllChem.MMFFOptimizeMolecule(mol)
    Chem.MolToXYZFile(mol, f'./vib2mol/SI/omp_demo_xyz/{smiles}.xyz')
    
```

# Naphthalene

```python
substructures = ['F', 'Cl', 'O', 'S', 'C=O', 'OC', 'C(O)=O', 'N', 'C#N', 'C#C', 'C=C', 'C', 'CC', 'CCC', '[N+](=O)[O-]', 'S(=O)(=O)O', 'P(O)(O)=O', 'C(N)=O', 'C(=O)C', 'C(Cl)(Cl)Cl', 'C(Cl)Cl', 'CCl', 'C(F)(F)F', 'C(F)F', 'CF']
len(substructures)
```

```python
import matplotlib.pyplot as plt
from rdkit import Chem
from rdkit.Chem import Draw
# ignore warnings
from rdkit import RDLogger
RDLogger.DisableLog('rdApp.error')
RDLogger.DisableLog('rdApp.warning')

# single substitution
single_substitution = []
substitute_type = []
substitute_substructure = []

backbones = {
    '2': 'c1{substructure}ccc2ccccc2c1', 
    '1': 'c1ccc2c{substructure}cccc2c1',
    }

for k, backbone in backbones.items():
    for substructure in substructures:
        tmp_smiles = backbone.format(substructure=f'({substructure})')
        mol = Chem.MolFromSmiles(tmp_smiles)
        single_substitution.append(Chem.MolToSmiles(mol, canonical=True, isomericSmiles=True, kekuleSmiles=True))
        substitute_type.append(k)
        substitute_substructure.append(substructure)

```

```python
# neighbor substitution
double_substitution = []
double_substitution_type = []
double_substitution_substructure = []
cache_inchi_list = []

backbones = {'2,6': 'c1{substructure1}ccc2cc{substructure2}ccc2c1', 
             '2,7': 'c1{substructure1}ccc2ccc{substructure2}cc2c1', 
             '1,5': 'c1ccc2c{substructure1}cccc2c1{substructure2}', 
             '1,8': 'c1ccc2cccc{substructure1}c2c1{substructure2}', 
             '1,2': 'c1{substructure1}ccc2ccccc2c1{substructure2}', 
            }


for substructure1 in substructures:
    for substructure2 in substructures:
        for k, backbone in backbones.items():
            tmp_smiles = backbone.format(substructure1=f'({substructure1})', substructure2=f'({substructure2})')
            mol = Chem.MolFromSmiles(tmp_smiles)
            tmp_inchi = Chem.MolToInchi(mol)
            if tmp_inchi in cache_inchi_list:
                continue
            cache_inchi_list.append(tmp_inchi)
            double_substitution.append(Chem.MolToSmiles(mol, canonical=True, isomericSmiles=False, kekuleSmiles=True))
            double_substitution_type.append(k)
            double_substitution_substructure.append([substructure1, substructure2])
```

```python
naphthaleme_substitution = single_substitution + double_substitution
naphthaleme_substitution_type = substitute_type + double_substitution_type
naphthaleme_substitution_substructure = substitute_substructure + double_substitution_substructure

(len(naphthaleme_substitution), len(naphthaleme_substitution_type), len(naphthaleme_substitution_substructure))
```

# Anthracene

```python
import matplotlib.pyplot as plt
from rdkit import Chem
from rdkit.Chem import Draw
# ignore warnings
from rdkit import RDLogger
RDLogger.DisableLog('rdApp.error')
RDLogger.DisableLog('rdApp.warning')

# single substitution
single_substitution = []
substitute_type = []
substitute_substructure = []

backbones = {'2': 'c1{substructure}ccc2cc3ccccc3cc2c1',
             '1': 'c1ccc2cc3ccccc3cc2c1{substructure}',
             '9': 'c1ccc2c{substructure}c3ccccc3cc2c1',
            }

for k, backbone in backbones.items():
    for substructure in substructures:
        tmp_smiles = backbone.format(substructure=f'({substructure})')
        mol = Chem.MolFromSmiles(tmp_smiles)
        single_substitution.append(Chem.MolToSmiles(mol, canonical=True, isomericSmiles=True, kekuleSmiles=True))
        substitute_type.append(k)
        substitute_substructure.append(substructure)

```

```python
# neighbor substitution
double_substitution = []
double_substitution_type = []
double_substitution_substructure = []

cache_inchi_list = []

backbones = {'2,3': 'c1{substructure1}c{substructure2}cc2cc3ccccc3cc2c1', # 2-3
             '2,6': 'c1{substructure1}ccc2cc3cc{substructure2}ccc3cc2c1', # 2-6
             '1,2': 'c1{substructure1}ccc2cc3ccccc3cc2c1{substructure2}', # 1-2
            }


for substructure1 in substructures:
    for substructure2 in substructures:
        for k, backbone in backbones.items():
            tmp_smiles = backbone.format(substructure1=f'({substructure1})', substructure2=f'({substructure2})')
            mol = Chem.MolFromSmiles(tmp_smiles)
            tmp_inchi = Chem.MolToInchi(mol)
            if tmp_inchi in cache_inchi_list:
                continue
            cache_inchi_list.append(tmp_inchi)
            double_substitution.append(Chem.MolToSmiles(mol, canonical=True, isomericSmiles=False, kekuleSmiles=True))
            double_substitution_type.append(k)
            double_substitution_substructure.append([substructure1, substructure2])
```

```python
antheracene_substitution = single_substitution + double_substitution
antheracene_substitution_type = substitute_type + double_substitution_type
antheracene_substitution_substructure = substitute_substructure + double_substitution_substructure
```

```python
(len(antheracene_substitution), len(naphthaleme_substitution), len(benzene_substitution))
```

```python
# from rdkit.Chem import AllChem
# # get optimal 3d conformations and output xyz files

all_substitutions = antheracene_substitution + naphthaleme_substitution
label_dict = {}

for i, inchi in enumerate(all_substitutions):
    try:
        mol = Chem.MolFromInchi(inchi)
        tmp_smiles = Chem.MolToSmiles(mol)
        mol = Chem.AddHs(mol)
        AllChem.EmbedMolecule(mol)
        AllChem.MMFFOptimizeMolecule(mol)
        Chem.MolToXYZFile(mol, f'./vib2mol/SI/pah_xyz/{i}.xyz')
        label_dict[i] = tmp_smiles
    except Exception as e:
        print(e)
        continue    
```
