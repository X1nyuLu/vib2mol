```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from adjustText import adjust_text

# ============================================================
# Font setup
# ============================================================
font_path = './appendix/Arial.ttf'
try:
    matplotlib.font_manager.fontManager.addfont(font_path)
    plt.rcParams['font.family'] = 'Arial'
except FileNotFoundError:
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Helvetica', 'Liberation Sans', 'DejaVu Sans']

plt.rcParams.update({
    'axes.spines.top':   False,
    'axes.spines.right': False,
    'axes.spines.left':  False,
})

# ============================================================
# Unified font-size system  (single source of truth)
# ============================================================
FS_TITLE   = 14   # panel titles  "(a) QM9S"
FS_AXIS    = 12   # axis labels   "Top-1 Recall (%)"
FS_TICK    = 10   # tick labels & y-group labels
FS_LEGEND  = 10   # legend text   (figure-level + in-panel)
FS_ANNOT   = 10    # data-value labels on bars
FS_SECTION = 9    # "Retrieval / Generation" italic annotations

# ============================================================
# 1. Data
# ============================================================
methods = ['VibraCLIP', 'PBSA', 'IR2Mol', 'SMEN', 'TranSpec', 'TranSpec (+formula)',
           'Vib2Mol', 'Vib2Mol (+formula)']
N = None

data_all = {
    'QM9S': {
        'VibraCLIP':          [97.07, 95.23, N,     N    ],
        'PBSA':               [N,     N,     78.77, 69.21],
        'IR2Mol':             [N,     N,     80.55, 75.39],
        'TranSpec':           [N,     N,     83.30, 76.70],
        'SMEN':               [97.89, 97.04, 65.78, 51.54],
        'Vib2Mol':            [97.18, 95.00, 89.02, 82.68],
        'Vib2Mol (+formula)': [98.11, 96.63, 90.91, 86.77],
        'TranSpec (+formula)': [N,     N,     N,     N    ],
    },
    'VB-Mols': {
        'VibraCLIP':          [91.21, 89.01, N,     N    ],
        'PBSA':               [N,     N,     69.37, 61.46],
        'IR2Mol':             [N,     N,     74.79, 71.44],
        'TranSpec':           [N,     N,     N,     N    ],
        'SMEN':               [95.43, 94.02, 53.08, 42.25],
        'Vib2Mol': [93.20, 91.38, 82.59, 78.59],
        'Vib2Mol (+formula)': [94.66, 93.38, 86.74, 83.69],
        'TranSpec (+formula)': [N,     N,     N,     N    ],
    },
    'SDBS': {
        'VibraCLIP':          [68.57, 74.28, N,     N    ],
        'PBSA':               [N,     N,     41.29, 30.68],
        'IR2Mol':             [N,     N,     50.71, 48.22],
        'TranSpec':           [N,     N,     N,     N    ],
        'SMEN':               [N,     N,     N,     N    ],
        'Vib2Mol':            [78.01, 71.28, 24.11, 24.82],
        'Vib2Mol (+formula)': [90.43, 86.17, 56.03, 52.84],
        'TranSpec (+formula)': [N,     N,     N,     N    ],
    },
    'NIST-IR': {
        'VibraCLIP':          [67.47, N    ],
        'PBSA':               [N,     41.56],
        'IR2Mol':             [N,     50.78],
        'TranSpec':           [N,     32.40],
        'TranSpec (+formula)':[N,     53.60],
        'SMEN':               [N,     N    ],
        'Vib2Mol':            [70.78, 34.07],
        'Vib2Mol (+formula)': [83.54, 55.14],
    },
}

task_labels_4 = ['Retrieval\n(Raman)', 'Retrieval\n(IR)', 'Generation\n(Raman)', 'Generation\n(IR)']
task_labels_2 = ['Retrieval\n(IR)',   'Generation\n(IR)']

colors = {
    'VibraCLIP':          '#fdd7b2',
    'PBSA':               '#d4b8e0',
    'IR2Mol':             '#F5F2E0',
    'TranSpec':           '#B3B3B3',
    'TranSpec (+formula)':'#7a7a7a',
    'SMEN':               '#E6DFB8',
    'Vib2Mol':            '#C4DEE2',
    'Vib2Mol (+formula)': '#8db7c3',
}

# ============================================================
# 2. Multi-modal data
# ============================================================
labels_mm = [
    'QM9S-IR', 'QM9S-Raman', 'QM9S-IR-Raman',
    'VB-Mols-IR', 'VB-Mols-Raman', 'VB-Mols-IR-Raman',
    'SDBS-IR', 'SDBS-Raman', 'SDBS-IR-Raman',
]
label_colors_mm = {
    'QM9S-IR': 'darkred',    'QM9S-Raman': 'darkred',    'QM9S-IR-Raman': 'darkred',
    'VB-Mols-IR': 'dimgray', 'VB-Mols-Raman': 'dimgray', 'VB-Mols-IR-Raman': 'dimgray',
    'SDBS-IR': 'darkblue',   'SDBS-Raman': 'darkblue',   'SDBS-IR-Raman': 'darkblue',
}
values_mm_ret = np.array([
    [96.63, 98.11, 98.74, 93.38, 94.66, 95.23, 86.17, 90.43, 94.33],
    [97.99, 98.46, 98.63, 93.01, 93.92, 94.68, 93.26, 93.26, 95.74],
])
values_mm_gen = np.array([
    [86.77, 90.91, 94.49, 83.69, 86.74, 89.24, 52.84, 56.03, 68.44],
    [90.77, 92.46, 93.79, 85.37, 86.94, 88.63, 60.28, 60.28, 67.02],
])
colors_mm = {'Vib2Mol': '#C4DEE2', 'Vib2Mol-MM': '#8db7c3'}

# ============================================================
# 3. Panel helper: per-dataset horizontal bars  (a–d)
# ============================================================
def plot_hbar_panel(ax, ds, title_text, xlim=None):
    
    data       = data_all[ds]
    ir_only    = (ds == 'NIST-IR')
    n_tasks    = 2 if ir_only else 4
    t_labels   = task_labels_2 if ir_only else task_labels_4

    bar_h    = 1
    gap      = 0.35
    centres  = []
    texts    = []
    
    for t in range(n_tasks):
        present = [m for m in methods if data[m][t] is not None]
        n_p     = len(present)
        centre  = t * (7 * bar_h + gap)
        centres.append(centre)
        for k, m in enumerate(present):
            v      = data[m][t]
            offset = (k - n_p / 2 + 0.5) * bar_h
            ax.barh(centre + offset, v, height=bar_h * 0.92,
                    color=colors[m], alpha=0.9, edgecolor='white', linewidth=1.2, zorder=3)
            
            txt = ax.text(v, centre + offset, f'{v:.1f}',
                    ha='left', va='center', fontsize=FS_ANNOT,
                    color='black', fontweight='bold',
                    bbox=dict(boxstyle='round,pad=0.12',
                              fc='lightgrey', ec='none', alpha=0.5),
                    zorder=4)
            
            texts.append(txt)
            
            
    # dashed separator between Retrieval and Generation
    if not ir_only:
        ax.axhline((centres[1] + centres[2]) / 2,
                   color='gray', linestyle=':', linewidth=1, alpha=0.7)

    # # section annotations
    # pairs = ([(centres[0], 'Retrieval'), (centres[1], 'Generation')]
    #          if ir_only else
    #          [((centres[0]+centres[1])/2, 'Retrieval'),
    #           ((centres[2]+centres[3])/2, 'Generation')])
    # for y_pos, label in pairs:
    #     ax.text(108, y_pos, label, ha='left', va='center',
    #             fontsize=FS_SECTION, style='italic', color='dimgray')

    ax.set_yticks(centres)
    ax.set_yticklabels(t_labels, fontsize=FS_TICK)
    ax.invert_yaxis()
    ax.set_xlabel('Top-1 Recall (%)', fontsize=FS_AXIS, fontweight='bold')

    if xlim:
        ax.set_xlim(xlim)
        ax.set_xticks(np.arange(xlim[0], xlim[1]+1, 20))  # 刻度也跟着变
    else:
        ax.set_xlim(0, 105)
        ax.set_xticks(np.arange(0, 101, 20))

    ax.tick_params(axis='x', labelsize=FS_TICK)
    ax.tick_params(axis='y', length=0)
    ax.set_title(title_text, loc='left', fontsize=FS_TITLE, fontweight='bold', pad=8)
    ax.grid(axis='x', linestyle='--', alpha=0.4, zorder=0)
    ax.set_axisbelow(True)

    for lbl in ax.get_yticklabels():
        lbl.set_fontweight('bold')

    # 所有 bar 画完后统一调整，只在 x 方向移动
    adjust_text(
        texts,
        ax=ax,
        only_move={'text': 'x', 'text': 'y'},          # 只横向推开，保持 y 与 bar 对齐
        autoalign=True,                      # 自动选择最佳推开方向
        arrowprops=dict(arrowstyle='-',   # 超出时画细引导线
                        color='gray',
                        lw=0.4),
        expand=(1.1, 1.1),                # 横向留 10% 余量
    )

# ============================================================
# 4. Panel helper: multi-modal horizontal bars  (e, f)
# ============================================================
def plot_mm_panel(ax, values_arr, title_text, xlim_left):
    bar_w  = 0.25
    idx    = np.arange(len(labels_mm))

    for i, (model, legend_label) in enumerate(
            [('Vib2Mol', 'w/o MMM'), ('Vib2Mol-MM', 'with MMM')]):
        vals = values_arr[i]
        ax.barh(idx + i * bar_w, vals, height=bar_w,
                label=legend_label, color=colors_mm[model],
                alpha=0.9, edgecolor='white', linewidth=1.2, zorder=3)
        for j, v in enumerate(vals):
            v_off = -0.06 if i == 0 else 0.06
            ax.text(v + 0.3, idx[j] + i * bar_w + v_off, f'{v:.1f}',
                    ha='left', va='center', fontsize=FS_ANNOT,
                    color='black', fontweight='bold',
                    bbox=dict(boxstyle='round,pad=0.12',
                              fc='lightgrey', ec='none', alpha=0.5),
                    zorder=4)

    ax.set_yticks(idx + bar_w / 2)
    ax.set_yticklabels(labels_mm, fontsize=FS_TICK)
    ax.invert_yaxis()
    for lbl in ax.get_yticklabels():
        lbl.set_color(label_colors_mm.get(lbl.get_text(), 'black'))
        lbl.set_fontweight('bold')

    ax.set_xlabel('Top-1 Recall (%)', fontsize=FS_AXIS, fontweight='bold')
    ax.set_xlim(xlim_left, 105)
    ax.tick_params(axis='x', labelsize=FS_TICK)
    ax.tick_params(axis='y', length=0)
    ax.grid(axis='x', linestyle='--', alpha=0.4, zorder=0)
    if '(f)' in title_text:
        ax.legend(fontsize=FS_LEGEND, frameon=True, loc='lower right')
    ax.set_title(title_text, loc='left', fontsize=FS_TITLE, fontweight='bold', pad=8)
    ax.spines['left'].set_visible(False)

# ============================================================
# 5. Figure & GridSpec
# ============================================================
fig = plt.figure(figsize=(12, 8), dpi=300, constrained_layout=True)
gs  = fig.add_gridspec(nrows=2, ncols=3,
                       width_ratios=[1, 1, 1],
                       height_ratios=[1, 1],
                       hspace=0.2, wspace=0.01)

for (r, c), ds, title in [
        ((0, 0), 'QM9S',    '(A) QM9S'),
        ((0, 1), 'VB-Mols', '(B) VB-Mols'),
        ((0, 2), 'SDBS',    '(C) SDBS'),
        ((1, 0), 'NIST-IR', '(D) NIST-IR'),
]:
    plot_hbar_panel(fig.add_subplot(gs[r, c]), ds, title, xlim=(0, 105))

plot_mm_panel(fig.add_subplot(gs[1, 1]), values_mm_ret,
              '(E) Multi-modal Retrieval',  xlim_left=85)
plot_mm_panel(fig.add_subplot(gs[1, 2]), values_mm_gen,
              '(F) Multi-modal Generation', xlim_left=50)

# ============================================================
# 6. Figure-level legend
# ============================================================
fig.legend(
    handles=[Patch(facecolor=colors[m], edgecolor='gray',
                   linewidth=0.4, label=m) for m in methods],
    loc='center', bbox_to_anchor=(0.5, 0.5),
    ncol=8, frameon=True, fontsize=FS_LEGEND,
)

# ============================================================
# 7. Save
# ============================================================
# plt.savefig('/mnt/user-data/outputs/fig_main_horizontal_v2.pdf',
#             dpi=300, bbox_inches='tight')
plt.savefig('figure3-v2.svg', dpi=300, bbox_inches='tight')

# print("saved.")
```

```python
import pandas as pd
import numpy as np

df_formula = pd.read_csv('appendix/formula_rerank.csv')
from rdkit import Chem

fg_dict = {
'hydroxyl':'[#6][OX2H]', 
'ketone carbonyl':'[#6][CX3](=O)[#6]', 
'aldehydes carbonyl':'[CX3H1](=O)[#6,H]', 
'carboxyl':'[CX3](=O)[OX2H]', 
'ester':'[#6][CX3](=O)[OX2H0][#6]', 
'amine':'[NX3;H2,H1,H0;!$(NC=O)]', 
'amide':'[NX3][CX3](=[OX1])[#6]', 
'ether':'[OD2]([#6])[#6]', 
'alkene':'[CX3]=[CX3]', 
'aromatic ring':'[$([cX3](:*):*),$([cX2+](:*):*)]', 
'alkyl halide':'[#6][F,Cl,Br,I]', 
'nitrile':'[NX1]#[CX2]', 
'thiol':'[#16X2H]',
'sulfide':'[#16X2H0]',
'alkyne':'[CX2]#[CX2]',
}

def classify_smiles(smiles):
    """
    classify a SMILES string into different functional groups.
    """
    mol = Chem.MolFromSmiles(smiles)
    correct = np.zeros(len(fg_dict), dtype=np.float32)

    for i, (fg, smarts) in enumerate(fg_dict.items()):
        pattern = Chem.MolFromSmarts(smarts)
        if mol and mol.HasSubstructMatch(pattern):
            correct[i] = 1.0
        else:
            correct[i] = 0.0
    return correct

def classify_smiles_list(smiles_list):
    """
    classify a list of SMILES strings into different functional groups.
    """
    classifications = []
    for smiles in smiles_list:
        classifications.append(classify_smiles(smiles))
    return classifications

df_formula['tgt_fg'] = classify_smiles_list(df_formula['tgt'])
df_formula['pred_fg'] = classify_smiles_list(df_formula['pred'])

for fg in fg_dict.keys():
    df_formula[fg] = df_formula['tgt_fg'].apply(lambda x: x[list(fg_dict.keys()).index(fg)])
```

```python
# evaluate the classification with F1 score of each functional group
from sklearn.metrics import f1_score

def evaluate_classification(df, fg_dict):
    """
    evaluate the classification results with F1 score of each functional group.
    """
    f1_scores = {}
    for fg in fg_dict.keys():
        y_true = df['tgt_fg'].apply(lambda x: x[list(fg_dict.keys()).index(fg)])
        y_pred = df['pred_fg'].apply(lambda x: x[list(fg_dict.keys()).index(fg)])
        f1_scores[fg] = f1_score(y_true, y_pred, zero_division=0)
    return f1_scores

def get_num_of_each_fg(df, fg_dict):
    """
    calculate the number of samples for each functional group.
    """
    num_of_each_fg = {}
    for fg in fg_dict.keys():
        num_of_each_fg[fg] = df['tgt_fg'].apply(lambda x: x[list(fg_dict.keys()).index(fg)]).sum()
    return num_of_each_fg


num_of_each_fg = get_num_of_each_fg(df_formula, fg_dict)
f1_scores_formula = evaluate_classification(df_formula, fg_dict)
```

```python
from rdkit import Chem
from rdkit.Chem import DataStructs
def taminoto_similarity(smiles1, smiles2):
    try:
        mol1 = Chem.MolFromSmiles(smiles1)
        mol2 = Chem.MolFromSmiles(smiles2)
        fp1 = Chem.RDKFingerprint(mol1, fpSize=2048)
        fp2 = Chem.RDKFingerprint(mol2, fpSize=2048)
        return DataStructs.TanimotoSimilarity(fp1, fp2)
    except:
        return 0

df_formula['tanimoto'] = df_formula.apply(lambda x: taminoto_similarity(x['pred'], x['tgt']), axis=1)
selected_df_gen = df_formula[df_formula.pred != '*']
selected_df_gen['tanimoto'].mean(), selected_df_gen[selected_df_gen.correct != 1]['tanimoto'].mean(), selected_df_gen[selected_df_gen.correct == 1]['tanimoto'].mean()
```
